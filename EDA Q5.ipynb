**1. Tighten Lending for Low CIBIL Borrowers**

***Evidence***

From EDA (Q2b), defaulted borrowers show a significantly lower CIBIL score distribution compared to performing borrowers, with a clear left shift in KDE curves. Mean CIBIL for defaulters is materially lower than non-defaulters.

***Business Impact***

Borrowers below the risky threshold (e.g., <650) contribute disproportionately to defaults, accounting for a large share of the ~77,556 default cases (3.88%). Tightening this segment can potentially reduce portfolio LGD exposure by 5–8% annually, saving several hundred crore in write-offs.

***Action***

Introduce stricter credit approval rules:

1. Auto-reject loans below CIBIL 650 OR
2. Require collateral or co-borrower for borderline applicants (650–700)

2. **Enforce EMI-to-Income Affordability Cap**

***Evidence***

From feature engineering (Q3a), EMI-to-income ratio showed a positive correlation with LGD. Higher ratios indicate repayment stress.

***Business Impact***

Loans with high EMI burden contribute disproportionately to default severity. Reducing exposure to high EMI-to-income borrowers can lower portfolio risk by an estimated 3–6% reduction in LGD exposure, improving overall credit stability.

***Action***

1. Set hard cap: EMI ≤ 35% of monthly income
2. Flag loans above 30% for manual underwriting review
3. Integrate EMI-to-income ratio into real-time loan approval system

3. **Revise Risk Weighting for High-Risk Loan Purposes**

***Evidence***

EDA (Q2g) showed strong variation in default rates by loan purpose, with certain categories consistently showing higher-than-average default rates (e.g., debt consolidation, small business loans).

***Business Impact***

High-risk loan purposes have up to 2–3× higher default rates compared to low-risk purposes, leading to disproportionate contribution to credit losses.

***Action***

1. Increase interest rates for high-risk purposes by 100–200 bps
2. Reduce approval rate for historically high-default purposes
3. Introduce purpose-based risk scoring overlay in underwriting

4. **Strengthen Monitoring for High-Risk Loan Grades (D–G)**

***Evidence***

EDA (Q2f) shows a monotonic increase in default rate from Grade A to Grade G, with a sharp jump in lower grades.

***Business Impact***

Lower-grade loans (D–G) contribute disproportionately to defaults despite smaller portfolio share. Tightening these grades can reduce portfolio defaults by an estimated 4–7% annually.

***Action***

1. Increase provisioning requirements for grades D–G
2. Reduce exposure limits for lower grades
3. Implement stricter post-disbursement monitoring for these segments

5. **Dynamic Pricing Based on Macro and Borrower Risk Signals**

***Evidence***

Q3 features and regression results show significant influence of:

1. Interest rate spread
2. COVID-year flag
3. Macroeconomic indicators (repo rate, inflation)

These variables directly impact LGD and borrower stress.

***Business Impact***

Dynamic pricing aligned with macro conditions can improve risk-adjusted returns by 2–4% annually, while reducing concentration risk during economic downturns.

***Action***

Introduce dynamic interest rate model tied to:
1. RBI repo rate changes
2. Inflation trends
3. Borrower risk score
4. Auto-adjust pricing bands quarterly
